## how too use Pydantic
1. Define a Pydantic model that represents the ideal schema of the data
    - this includes the expected fields, their types, and any validation constraints (e.g `gt=0` for positve numbers).
2. instantiate the model with raw input data (usually a dictionary or JSON-like structure).
    - Pydantic will automatically validate the data and coerce it into the correct Python types (if possible).
    - if the data doesn't meet the model's requirements, Pydantic rasis a `ValidationError`
3. Pass the validated model object to functions or use it throughout our codebase.
    - this ensures that every part of your program works with clean, type-safe, and logically valid data.

## compute feild

to automatically calculate something on add use `@computed_field` and mayhaps `@property` inside the class

```python
class Patient(BaseModel):
    name: str = Field(..., description="The name of the patient")
    age: int
    height: float
    weight: float
    married: bool = False
    allergy: List[str] = []
    contact_info: Dict[str, str] = {}
    
    @computed_field
    @property
    def bmi(self) -> float:
        """Calculate and return the Body Mass Index (BMI) of the patient."""
        if self.height <= 0:
            raise ValueError("Height must be greater than zero.")
        return self.weight / (self.height ** 2)
    
Patient_data = {
    "name": "John Doe",
    "age": 30,
    "height": 1.75,  # height in meters
    "weight": 70.0,  # weight in kilograms
    "married": True,
    "allergy": ["peanuts"],
    "contact_info": {"email": "john.doe@example.com"}
}

patient = Patient(**Patient_data)
print(f"Patient {patient.name} has a BMI of {patient.bmi:.2f}.")
```
