```python
@agent.tool_plain
def calculate_damage(attack: int, defense: int) -> int:
    """Calculate damage dealt by an attack."""
    print(f"TOOL EXECUTED: attack={attack}, defense={defense}")
    return max(attack - defense, 0)
```

### NOTE:
- `@agent.tool_plain`: this registers this function as a tool for the agent.
    the decorator is the important part.

```python
result = await agent.run(
    """
    An attacker has 10 attack.
    A defender has 3 defense.

    Use the calculate_damage tool.
    Then state the resulting damage.
    """
)

print(result.output)
```
