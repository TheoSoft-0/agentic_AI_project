```python
import sys

from pathlib import Path
from time import sleep

PROJECT_ROOT = Path.cwd()
SRC_ROOT = PROJECT_ROOT / "src"

if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))
    
from agentic_ai.agent import (
    ProtonAgentDeps,
    create_proton_agent,
)
from agentic_ai.proton import (
    close_proton_mail,
    ensure_logged_in,
    open_proton_mail,
    open_trash,
)

```

```python
playwright, context, page = await open_proton_mail(PROJECT_ROOT)

await ensure_logged_in(page)
await page.wait_for_timeout(5000)
await open_trash(page)
```

```python
email_agent = create_proton_agent()

deps = ProtonAgentDeps( page=page, allow_writes=False,)
```

```python
result = await email_agent.run(
    """
Inspect Trash conversation 0.

Classify it as ignore, review, or urgent.

Do not modify the conversation.
Return the classification and your reason.
""",
    deps=deps,
)

print(result.output)
print()
print(result.all_messages())
```
