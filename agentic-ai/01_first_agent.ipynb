```python
import pydantic_ai

print(pydantic_ai.__version__)
```

```python
from pydantic_ai import Agent
from pydantic_ai.models.ollama import OllamaModel
from pydantic_ai.providers.ollama import OllamaProvider


model = OllamaModel(
    "qwen3:4b-instruct",
    provider=OllamaProvider(
        base_url="http://localhost:11434/v1"
    ),
)

agent = Agent(model)
```

## NOTE:
- `OllamaModel(...)`: 

    tells Pydantic AI to use the Qwen model running throuth Ollama.



- `OllamaProvider(...)`:

    tells it the Ollama server is located at base_url=...

- `Agent(model)`:

    creates the actual agent object.

```python
result = await agent.run(
    "What is 2 + 2? Answer in one sentence."
)

print(result.output)
```
